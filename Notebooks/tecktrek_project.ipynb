Even though these products have missing details, people still order them, so its best not to remove these entries from the dataframe

```python
def object_clean(df):
    for col in df.columns:
        if df[col].dtype == 'object':
            df[col] = df[col].str.strip()
            df[col] = df[col].astype('category')
    return df
```

```python
customers_df = object_clean(customers_df)
geo_location_df = object_clean(geo_location_df)
order_items_df = object_clean(order_items_df)
order_payments_df = object_clean(order_payments_df)
order_reviews_df = object_clean(order_reviews_df)
orders_df = object_clean(orders_df)
products_df = object_clean(products_df)
sellers_df = object_clean(sellers_df)
product_translation_df = object_clean(product_translation_df)
```
